### 5. Create a new dataframe from the raw/uncleaned data called *‘fatality_locations’* with the following columns:
### *‘Date’, ‘Location’, ‘Aboard’, ‘Fatalities’.*

```python
# Create a new dataframe containing only the required columns
fatality_locations = df[['Date', 'Location', 'Aboard', 'Fatalities']].copy()

print("The fatality_locations dataframe is:")
display(fatality_locations)
```

### 6. Determine the date of the highest number of recorded fatalities from this new dataframe.

```python
max_fatalities_row = fatality_locations.loc[fatality_locations['Fatalities'].idxmax()]

print(f"Date of highest fatalities: {max_fatalities_row['Date']}")
print(f"Location: {max_fatalities_row['Location']}")
print(f"Fatalities: {max_fatalities_row['Fatalities']}")
```

### 7. Compare the number of passengers aboard to the number of fatalities per crash. Are there any recorded crashes where there were no fatalities? If so, how many?

```python
# Display the number of people aboard and the number of fatalities for each crash
print("Number of people aboard compared to fatalities:")
display(fatality_locations[['Aboard', 'Fatalities']])

# Count the number of crashes where there were no fatalities
no_fatalities = (fatality_locations['Fatalities'] == 0).sum()

print(f"Number of crashes with no fatalities: {no_fatalities}")
```

### 8. In the *‘fatality_locations’* dataframe, the location column has data items represented as ‘Region, U.S. State/Country’. Split this column to have two columns of regions and states/countries.

```python
# The str.split() function splits the Location column at the comma
# n=1 ensures that the split happens only at the first comma
fatality_locations[['Region', 'State/Country']] = fatality_locations[
    'Location'
].str.split(',', n=1, expand=True)

print("The Location column has been split into Region and State/Country:")
display(fatality_locations)
```

### 9. Order the dataframe by the number of fatalities from highest to lowest and select the first 100.

```python
# sort_values() orders the dataframe based on the Fatalities column
# ascending=False arranges the values from highest to lowest
fatality_locations = fatality_locations.sort_values(
    by='Fatalities',
    ascending=False
)

# Select the first 100 rows
top_100 = fatality_locations.head(100)

print("Top 100 crashes with the highest number of fatalities:")
display(top_100)
```

### 10. Generate a pie chart that shows the distribution of the top 25 number of fatalities per country/U.S. State. [Hint: charts can be generated from the seaborn library]

```python
# Group the crashes by State/Country and calculate the total fatalities
fatalities_by_country = fatality_locations.groupby(
    'State/Country'
)['Fatalities'].sum()

# Sort the total fatalities from highest to lowest
fatalities_by_country = fatalities_by_country.sort_values(
    ascending=False
)

# Select the top 25 countries/U.S. States
top_25 = fatalities_by_country.head(25)

print("Top 25 countries/U.S. States by number of fatalities:")
print(top_25)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a pie chart showing the distribution of fatalities
plt.figure(figsize=(10, 10))

plt.pie(
    top_25,
    labels=top_25.index,
    autopct='%1.1f%%'
)

plt.title('Distribution of Top 25 Fatalities by Country/U.S. State')
plt.show()
```
